```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from google.colab.patches import cv2_imshow #function imshow() para colab
```

```python
arquivo = files.upload()
```

```python
img = cv2.imread('rick-and-morty.jpg')

img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
image_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
cv2_imshow(img_rgb)
```

```python
plt.figure(figsize=(10,3))
plt.imshow(img_rgb)
plt.title('Rick and morty')
plt.axis('off')
plt.show()
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
fig, ax = plt.subplots(1, 3, figsize=(6, 4))
ax[0].axis('off')
ax[0].imshow(img)
ax[0].set_title('bgr')
ax[1].axis('off')
ax[1].imshow(img_rgb)
ax[1].set_title('rgb')
ax[2].imshow(image_hsv)
ax[2].axis('off')
ax[2].set_title('hsv')

plt.tight_layout()
plt.show()

```

```python
img_grey = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

fig, ax = plt.subplots(2, 2, figsize=(6, 4))
ax[0, 0].axis('off')
ax[0, 0].imshow(img)
ax[0, 0].set_title('bgr')

ax[0, 1].axis('off')
ax[0, 1].imshow(img_rgb)
ax[0, 1].set_title('rgb')

ax[1, 0].imshow(image_hsv)
ax[1, 0].axis('off')
ax[1, 0].set_title('hsv')

ax[1, 1].imshow(img_grey, cmap='gray')
ax[1, 1].axis('off')
ax[1, 1].set_title('grey')

plt.tight_layout()
plt.savefig('rick-and-morty-4')
plt.show()
```
