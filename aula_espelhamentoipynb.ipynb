```python
img_bgr = cv2.imread('oh my god.jpg')
```

```python
img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)
```

```python
img_gray
```

```python
altura = img_gray.shape[0]
largura= img_gray.shape[1]
```

```python
print(f'{largura} e {altura}')
```

```python
dx = 50
dy = 30
```

```python
altura_img = img_gray.shape[0]
largura_img = img_gray.shape[1]

transladada = np.zeros((altura_img, largura_img), dtype=np.uint8)

for y in range(altura_img):
  for x in range(largura_img):
    new_y = y + dy
    new_x = x + dx

    if (new_y >= 0 and new_y < altura_img):
      if (new_x >= 0 and new_x < largura_img):
        transladada[new_y, new_x] = img_gray[y, x]

plt.figure(figsize=(7,5))
plt.imshow(transladada, cmap='gray')
plt.axis('off')
plt.show()
```

```python
wrap_around = np.zeros((altura_img, largura_img), dtype=np.uint8)
dx = 300
dy = 150
for y in range(altura_img):
  for x in range(largura_img):
    new_y = (y + dy) % altura
    new_x = (x + dx) % largura

    wrap_around[new_y, new_x] = img_gray[y, x]




plt.figure(figsize=(7,5))
plt.imshow(wrap_around, cmap='gray')
plt.axis('off')
plt.show()
```

```python
sx =  0.1
sy = 0.1

nova_largura = int(largura * sx)
nova_altura  = int(altura*sy)



escalada = np.zeros((nova_altura, nova_largura), dtype=np.uint8)

for x in range(nova_largura):
  for y in range(nova_altura):
    new_x = int(x / sx)
    new_y = int(y/sy)
    escalada[y, x] = img_gray[new_y, new_x]


plt.figure(figsize=(7,5))
plt.imshow(escalada, cmap='gray')
plt.axis('off')
plt.show()
```

```python
## espelhamento
espelhado = np.zeros((altura, largura), dtype=np.uint8)

for x in range(nova_largura):
  for y in range(nova_altura):
    new_x = largura - 1 - x
    new_y = altura - 1 - y
    espelhado[new_y, new_x] = img_gray[y,x]

plt.figure(figsize=(7,5))
plt.imshow(espelhado, cmap='gray')
plt.axis('off')
plt.show()
```
